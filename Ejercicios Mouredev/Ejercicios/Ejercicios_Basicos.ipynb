```python
EXPECTED_BAKE_TIME = 40
PREPARATION_TIME = 2
def bake_time_remaining(elapsed_bake_time):
    '''
    :param elapsed_bake_time: int baking time already elapsed
    :return: int remaining bake time derived from 'EXPECTED_BAKE_TIME'
 
    Function that takes the actual minutes the lasagna has been in the oven as
    an argument and returns how many minutes the lasagna still needs to bake
    based on the `EXPECTED_BAKE_TIME`.
    '''
    return EXPECTED_BAKE_TIME - elapsed_bake_time
def preparation_time_in_minutes(number_of_layers):
    '''
    :param number_of_layers: int number of layers in the lasagna
    :return: int time to prepare the lasagna
 
    Function that calculates the time of preparation according to the number of
    layers.
    '''
    return PREPARATION_TIME * number_of_layers
def elapsed_time_in_minutes(number_of_layers, elapsed_bake_time):
    '''
    :param number_of_layers: int number of layers in the lasagna
    :param elapsed_bake_time: int baking time already elapsed
    :return: int baking time elapsed in minutes
 
    Function that calculates the elapsed time in minutes according to the number
    of layers and elapsed bake time.
    '''
    return preparation_time_in_minutes(number_of_layers) + elapsed_bake_time
```

```python

from typing import Union

def is_criticality_balance( temperature: Union[int, float], neutrons_emitted: Union[int, float])-> bool:
   
   """
   Verificar que la criticidad esté equilibrada.

:param temperature: int o float - valor de temperatura en kelvin.
:param neutrons_emitted: int o float - número de neutrones emitidos por segundo.
:return: bool - ¿la criticidad está equilibrada?

Se dice que un reactor es crítico si cumple las siguientes condiciones:
- La temperatura es inferior a 800 K.
- El número de neutrones emitidos por segundo es superior a 500.
- El producto de la temperatura por los neutrones emitidos por segundo es inferior a 500000.
   
   """ 
   return temperature < 800 and neutrons_emitted > 500 and temperature * neutrons_emitted < 500000



def reactor_efficiency (voltage : Union[int, float], current: Union[int, float], theoretical_max_power: Union[int, float])-> bool:
    
    """
    Evaluar la zona de eficiencia del reactor.

:param voltage: int o float - valor de voltaje.
:param current: int o float - valor de corriente.
:param theory_max_power: int o float - potencia que corresponde a una eficiencia del 100%.
:return: str - uno de ('verde', 'naranja', 'rojo' o 'negro').

La eficiencia se puede agrupar en 4 bandas:

1. verde -> eficiencia del 80% o más,
2. naranja -> eficiencia de menos del 80% pero al menos del 60%,
3. rojo -> eficiencia por debajo del 60%, pero aún del 30% o más,
4. negro -> eficiencia inferior al 30%.

El valor porcentual se calcula como
(potencia generada/potencia máxima teórica)*100
donde potencia generada = voltaje * corriente
    
    """

    generated_power = voltage  * current
    efficiency = generated_power / theoretical_max_power * 100


    if efficiency >= 80: 
        return "green"

    elif efficiency >= 60:
        return "orange"

    elif efficiency >= 30:
        return ' red'
    
    else:
        return " blak"
    


def fail_safe(temperature: Union[int,float], neutrons_produced_per_second:Union[int, float], threshold: Union[int, float]):
    """ 
    Evaluar y devolver el código de estado del reactor.

:param temperature: int o float - valor de la temperatura en kelvin.
:param neutrons_produced_per_second: int o float - flujo de neutrones.
:param threshold: int o float - umbral para la categoría.
:return: str - uno de ('LOW', 'NORMAL', 'DANGER').

1. 'LOW' -> `temperature * neutrons per second` < 90% de `threshold`
2. 'NORMAL' -> `temperature * neutrons per second` +/- 10% de `threshold`
3. 'DANGER' -> `temperature * neutrons per second` no está en los rangos indicados anteriormente
    """
    
    if threshold > 90 * 100:
        return 'LOW'
    elif threshold == 10 * 100:
        return 'NORMAL'
    else:
        return 'DANGER'

print(is_criticality_balance( temperature=750, neutrons_emitted=600) )

print(reactor_efficiency(voltage=200,current=50,theoretical_max_power=15000))

print(fail_safe(temperature=1000,neutrons_produced_per_second=30,threshold=5000))





```

```python
# Blackjack

from typing import Union

def value_of_card(card: str)-> int:
    
    ''' 
    Determinar el valor de puntuación de una carta.

    :param card: str: carta dada.
    :return: int: valor de una carta dada. Ver valores a continuación.

    1.  'J', 'Q', or 'K' (tambien conocidas como cartas con figuras) = 10
    2.  'A' ( carta As ) = 1
    3.  '2' - '10' = valor de cada carta.

'''
    if card in 'JKQ':
        return 10
    elif card == 'A':
        return 1
    return int(card)
    

value_of_card("4")

def higher_card(card_one: str, card_two: str)-> str:
    
    """ 
    Determina qué carta tiene un valor más alto en la mano.

    :param card_one, card_two: str: cartas repartidas en la mano. Ver los valores a continuación.
    :return: str o tuple: la tupla resultante contiene ambas cartas si tienen el mismo valor.
    """
    if value_of_card(card_one) > value_of_card(card_two):
        return card_one
    elif value_of_card(card_one) < value_of_card(card_two):
        return card_two
    return card_one,card_two


higher_card("2", "10")
higher_card("6","5")


def value_of_ace(card_one: str, card_two: str)-> int:
    ''' 
    Calcula el valor más ventajoso para la carta as.

    :param card_one, card_two: str - carta repartida. Consulta los valores a continuación.
    :return: int - valor 1 u 11 de la próxima carta as.

    1. 'J', 'Q' o 'K' (también conocidas como "cartas con figuras") = 10
    2. 'A' (carta as) = ​​11 (si ya está en la mano)
    3. '2' - '10' = valor numérico.
    
    '''
    if card_one == "A" or card_two == "A" or value_of_card(card_one) + value_of_card(card_two) > 10:
        return 1
    return 11

value_of_ace( card_one="Q", card_two= "A" )



def is_blackjack(card_one : str, card_two: str)-> bool:
    
    """ 
    Determina si la mano es "natural" o "blackjack".

    : parámetro carta uno, carta dos: estrella: carta repartida. Consulta los valores a continuación. 
    : retorno: bool: la mano es un blackjack (dos cartas que valen 21)
    """
    if card_one == "A":
        if 11 + value_of_card(card_two)== 21:
            return True
    
    if card_two == "A":
        if value_of_card(card_one) + 11 == 21:
            return True
    
    return False

is_blackjack("9","10")
 
 
def can_split_pairs(card_one: str, card_two: str)-> bool:
 
 """ 
 Determinar si un jugador puede dividir su mano en dos manos.

:param card_one, card_two: str - cartas repartidas.
:return: bool - ¿se puede dividir la mano en dos pares? (es decir, las cartas tienen el mismo valor).

 """
 if value_of_card(card_one) == value_of_card(card_two):
  return True
 return False


def can_double_down(card_one: str, card_two: str)-> bool:

 """ 
    Determinar si un jugador de blackjack puede realizar una apuesta de doblar la apuesta.

    :param card_one, card_two: str - primera y segunda carta en la mano.
    :return: bool - ¿se puede doblar la apuesta en la mano? (es decir, el total es 9, 10 u 11 puntos)

"""
 if value_of_card(card_one) + value_of_card(card_two) in (9, 10, 11):
    return True
 return False
```
