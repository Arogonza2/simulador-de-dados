```python
def steps():
    number = int(input(" Escribe un numero mayor de 1"))  
    
    number_steps = 0  
    
    while number != 1:  
        if number %2 == 0:
            number = number// 2
            print(number)
        
        elif number %2 != 0:
            number = number *3 + 1
            print(number)
        
        number_steps += 1
    
    return number
    
steps()

```
